```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/kaggle/input/datasets/fantasticboy123/assign-ml/dataset.csv")

df = df.drop(columns=["Unnamed: 0"], errors="ignore")

# Table
table = df.groupby('Market_T')['Success_Percentage'].mean().reset_index()
table.to_csv("/kaggle/working/table_RQ6.csv", index=False)

# Plot
plt.figure(figsize=(7,5))
plt.plot(table['Market_T'], table['Success_Percentage'], marker='o')
plt.title("Market Trends vs Success Percentage")
plt.xlabel("Market Trend")
plt.ylabel("Success Percentage")

# Save
plt.savefig("/kaggle/working/figure_RQ6.pdf")
plt.show()
```
