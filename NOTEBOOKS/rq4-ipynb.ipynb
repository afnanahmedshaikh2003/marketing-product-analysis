```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/kaggle/input/datasets/fantasticboy123/assign-ml/dataset.csv")

df = df.drop(columns=["Unnamed: 0"], errors="ignore")

# Create bins
df['Rating_Bin'] = pd.cut(df['Rating'], bins=[0,2,4,5], labels=['Low','Medium','High'])

# Table
table = df.groupby('Rating_Bin')['Success_Percentage'].mean().reset_index()
table.to_csv("/kaggle/working/table_RQ4.csv", index=False)

# Plot
plt.figure(figsize=(7,5))
plt.plot(table['Rating_Bin'], table['Success_Percentage'], marker='o')
plt.title("Rating vs Success Percentage")
plt.xlabel("Rating Level")
plt.ylabel("Success Percentage")

# Save
plt.savefig("/kaggle/working/figure_RQ4.pdf")
plt.show()
```
