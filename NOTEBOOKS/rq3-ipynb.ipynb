```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("/kaggle/input/datasets/fantasticboy123/assign-ml/dataset.csv")

df = df.drop(columns=["Unnamed: 0"], errors="ignore")

# Table
table = df.groupby('Category')[['Success_Percentage','Sales_m']].mean().reset_index()
table.to_csv("/kaggle/working/table_RQ3.csv", index=False)

# Plot
plt.figure(figsize=(8,5))
sns.barplot(x='Category', y='Success_Percentage', data=table)
plt.xticks(rotation=45)
plt.title("Category vs Success Percentage")

# Save
plt.savefig("/kaggle/working/figure_RQ3.pdf")
plt.show()
```
