```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("/kaggle/input/datasets/fantasticboy123/assign-ml/dataset.csv")

df = df.drop(columns=["Unnamed: 0"], errors="ignore")

# Create bins
df['Efficiency_Level'] = pd.qcut(df['Supply_Chain_E'], 3, labels=['Low','Medium','High'])

# Table
table = df.groupby('Efficiency_Level')['Sales_m'].mean().reset_index()
table.to_csv("/kaggle/working/table_RQ5.csv", index=False)

# Plot
plt.figure(figsize=(7,5))
sns.boxplot(x='Efficiency_Level', y='Sales_m', data=df)
plt.title("Supply Chain Efficiency vs Sales")

# Save
plt.savefig("/kaggle/working/figure_RQ5.pdf")
plt.show()
```
